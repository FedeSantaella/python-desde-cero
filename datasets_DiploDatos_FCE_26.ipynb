## Dataset 3: Encuesta Permanente de Hogares (EPH)

image.png

Caracterización: La Encuesta Permanente de Hogares (EPH) es un programa nacional de producción sistemática y permanente de indicadores sociales que lleva a cabo el Instituto Nacional de Estadística y Censos (INDEC).
Con los datos obtenidos se proporcionan regularmente, las tasas oficiales de empleo, desocupación, subocupación y pobreza.
Es una encuesta por muestreo, lo que significa que para conocer las diversas características del total de los hogares, se encuesta una pequeña fracción representativa de los mismos. Cubre 31 aglomerados urbanos de más de 50 mil habitantes.



Para más información: https://www.indec.gob.ar/indec/web/Institucional-Indec-BasesDeDatos

```python
import pandas as pd
from zipfile import ZipFile
import requests
from io import BytesIO
import seaborn as sns
import matplotlib.pyplot as plt
from  matplotlib.ticker import FuncFormatter

# URL del archivo ZIP
url_zip = "https://www.indec.gob.ar/ftp/cuadros/menusuperior/eph/EPH_usu_4_Trim_2023_txt.zip"

# Descargar el archivo ZIP
response = requests.get(url_zip)

# Leer el archivo ZIP
with ZipFile(BytesIO(response.content)) as zip_file:
    # Listar los archivos dentro del ZIP
    files = zip_file.namelist()
    print("Archivos en el ZIP:", files)

    # Seleccionar el archivo deseado
    selected_file = files[2]

    # Extraer y leer el archivo CSV seleccionado
    with zip_file.open(selected_file) as csv_file:
        EPH_df = pd.read_csv(csv_file, sep=";")

# Mostrar las primeras filas del DataFrame
print(EPH_df.head(10))
```

El Boletín Estadístico de la Seguridad Social (BESS) es una publicación trimestral que brinda series estadísticas consolidadas, actualizadas y lo suficientemente dinámicas sobre las variables más relevantes de la seguridad social en la Argentina.

En dicho boletín se presentan estadísticas de las distintas ramas de la seguridad social: jubilaciones y pensiones, asignaciones familiares y prestación por desempleo, elaboradas por la Subsecretaría de Seguridad Social en base a la información proporcionada por la Agencia de Regulación y Control Aduanero (ARCA) y la Administración Nacional de la Seguridad Social (ANSES).


Para más información: https://www.argentina.gob.ar/trabajo/seguridadsocial/bess


```python
import pandas as pd
import requests
from io import BytesIO

# URL del archivo Excel BESS
url_bess = "https://www.argentina.gob.ar/sites/default/files/boletin_provincias_pasivos_05-2025.xlsx"

# Headers para evitar error 403
headers = {
    "User-Agent": "Mozilla/5.0"
}

# Descargar archivo
response = requests.get(url_bess, headers=headers)

# Verificar estado
print("Status:", response.status_code)

# Leer archivo Excel
xls = pd.ExcelFile(BytesIO(response.content))

# Ver hojas disponibles
print("\nHojas disponibles:")
print(xls.sheet_names)

# Leer primera hoja
bess = pd.read_excel(
    BytesIO(response.content),
    sheet_name=0
)

# Mostrar primeras filas
print("\nPrimeras filas:")
print(bess.head(10))

# Mostrar columnas
print("\nColumnas:")
print(bess.columns)

# Información general
print("\nInfo:")
print(bess.info())
```

Incluye información como ubicación (provincia, ciudad, barrio, coordenadas), precio, tipo de propiedad, tipo de operación (venta o alquiler), superficie, número de habitaciones y más.

**OBJETIVO**

*Facilitar* el análisis del mercado inmobiliario y la predicción de precios de venta o alquiler

**VARIABLES**

- `id`: Identificador único de cada publicación del inmueble en el dataset.  
- `ad_type`: Tipo de anuncio, normalmente indica si se trata de una propiedad.  
- `start_date`: Fecha de inicio de publicación del aviso inmobiliario.  
- `end_date`: Fecha de finalización del aviso; si aparece una fecha lejana (ej. 9999-12-31), suele indicar que sigue activo.  
- `created_on`: Fecha en la que el registro fue creado en la base de datos.  
- `lat`: Latitud de la ubicación geográfica del inmueble.  
- `lon`: Longitud de la ubicación geográfica del inmueble.  
- `l1`: Nivel 1 de localización (generalmente el país, por ejemplo Argentina).  
- `l2`: Nivel 2 de localización (provincia o ciudad principal, por ejemplo Capital Federal).  
- `l3`: Nivel 3 de localización (zona o barrio específico, por ejemplo Colegiales o Almagro).  
- `bathrooms`: Número de baños que tiene la propiedad.  
- `surface_total`: Superficie total del inmueble en metros cuadrados (incluye áreas cubiertas y descubiertas).  
- `surface_covered`: Superficie cubierta del inmueble en metros cuadrados.  
- `currency`: Moneda en la que está expresado el precio (por ejemplo USD o ARS).  
- `price_period`: Periodicidad del precio (por ejemplo mensual en caso de alquiler; en ventas suele estar vacío).  
- `title`: Título del anuncio inmobiliario.  
- `description`: Descripción detallada del inmueble proporcionada en el aviso.  
- `property_type`: Tipo de propiedad (por ejemplo departamento, casa, PH, terreno, etc.).  
- `operation_type`: Tipo de operación inmobiliaria (venta o alquiler).  
- `price`: Precio de la propiedad (variable objetivo principal para análisis y modelado).

**INTRODUCCION:**
Una de las principales actividades que realizan las entidades financieras es la gestión de transferencias y pagos digitales. Estas operaciones pueden verse afectadas por intentos de fraude que generan pérdidas económicas y riesgos para la seguridad de los clientes. Para la gestión de este riesgo, la entidad ha decidido utilizar Machine Learning. Se recopilaron datos históricos de transacciones simuladas que permiten el desarrollo de un modelo de ML para clasificar si una nueva operación es probablemente fraudulenta o legítima.




**OBJETIVO**

El objetivo del proyecto es, a través de los distintos logs (observaciones), que son transacciones sintéticas de dinero llevadas a cabo a través del móvil, detectar cuáles pueden ser fraudulentas.

Para poder realizar esto se deberá pasar por los distintos ciclos de la extracción de conocimiento (preparar los datos, minería de datos, visualización, etc), para conseguir la suficiente comprensión y estructura deseada de estos.

En resumen, los objetivos del proyecto son:

Estudiar el dataset y decidir cuáles son las técnicas a emplear
Aplicar técnicas de manipulación de datos necesarias (imputación, transformación, normalización, etc)
Visualización de datos con distintas gráficas
Detección de fraudes (Detectar correctactamente aquellos movimientos financieros fraudulentos)

**VARIABLES**

- `step`: Unidad de tiempo de la simulación. Cada step representa una hora desde el inicio del dataset. Cantidad total de steps 744 (30 días)  
- `type`: Tipo de transacción realizada. Puede tomar valores como `PAYMENT`, `TRANSFER`, `CASH_OUT`, `CASH_IN` o `DEBIT`.  
- `amount`: Importe monetario transferido en la operación.  
- `nameOrig`: Identificador único de la cuenta o cliente que origina la transacción (emisor).  
- `oldbalanceOrg`: Saldo de la cuenta origen antes de realizar la transacción.  
- `newbalanceOrig`: Saldo de la cuenta origen después de realizar la transacción.  
- `nameDest`: Identificador único de la cuenta o cliente destino que recibe la transacción.  
- `oldbalanceDest`: Saldo de la cuenta destino antes de recibir la transacción.  
- `newbalanceDest`: Saldo de la cuenta destino después de recibir la transacción.  
- `isFraud`: Variable objetivo que indica si la transacción fue identificada como fraudulenta (`1`) o legítima (`0`).  
- `isFlaggedFraud`: Indicador de alerta automática del sistema. Vale `1` cuando la transacción fue marcada como sospechosa por reglas internas, normalmente transferencias de gran importe.
